## Para qué sirve una revisión

Una revisión de código (*code review*) es leer el cambio de alguien **antes**
de que entre a la rama principal. Sirve para tres cosas:

- **Atrapar errores** que quien escribió ya no ve, porque sabe lo que quiso
  decir y lee eso.
- **Hacer el código legible para alguien más.** Si quien revisa no entiende una
  celda, la siguiente persona tampoco la va a entender.
- **Aprender**, en las dos direcciones. Quien revisa ve otra forma de resolver
  el mismo problema.

SymPy funciona así: ningún cambio entra a su rama principal sin que otra
persona lo revise en un PR. Este repositorio del curso también. El notebook que
tienes abierto llegó a `main` por un PR que alguien revisó.

## Anatomía de un PR

![Las pestañas de un Pull Request, un comentario de línea con sugerencia y el botón Review changes](img/anatomia-de-un-pr.svg)

Abre cualquier PR tuyo de entrega para seguir el recorrido. GitHub está en
inglés:

- **Conversation**: la discusión general y el resumen de cada revisión.
- **Commits**: la lista de commits de la rama. Ahí se nota si el historial
  cuenta una historia o es un solo commit gigante.
- **Checks**: el resultado del autograding de GitHub Actions.
- **Files changed**: el **diff**. Es la pestaña donde se revisa.

En *Files changed*, al pasar el cursor por una línea aparece un **`+`** azul.
Ese botón abre un **comentario de línea**: queda pegado a esa línea exacta, no
flotando en la conversación. *Start a review* junta varios comentarios y los
publica todos juntos al final con **Review changes**, que da tres opciones:

| Opción | Significa | Cuándo usarla hoy |
|---|---|---|
| *Comment* | opino, sin decidir | **siempre**: tú no calificas la tarea |
| *Approve* | está listo para fusionar | en tu proyecto, cuando lo revise tu pareja |
| *Request changes* | hay que corregir antes de fusionar | en tu proyecto, si hay un error real |

## Leer el diff de un notebook

Un `.ipynb` es un archivo JSON. Gracias a `nbstripout` (semana 3) tus
notebooks se versionan **sin salidas**, así que el diff solo muestra las líneas
de código y de Markdown que cambiaron, cada una entre comillas y con un `\n`
al final. Sin `nbstripout` el diff estaría lleno de imágenes codificadas y
números de ejecución, y nadie podría revisarlo.

GitHub también muestra los notebooks de forma "rica" (las celdas como en
Jupyter), con el botón de documento junto al nombre del archivo. Para comentar
una línea concreta conviene la vista de texto.

## Qué buscar en un notebook de física

Una lista de cotejo para no revisar a ciegas. No todas aplican a todos los PRs.

| Pregunta | Por qué importa |
|---|---|
| ¿Corre de punta a punta desde un kernel limpio? | si no, nada de lo demás se puede creer |
| ¿Los símbolos tienen las suposiciones que su física exige? | sin ellas, SymPy no se atreve a simplificar raíces ni valores absolutos |
| ¿Hay flotantes sueltos (`0.5`, `9.8`)? | contaminan la expresión entera y rompen la simplificación exacta |
| ¿Las expresiones se muestran con `display` y no con `print`? | `print` tira el renderizado matemático |
| ¿El resultado se **verifica** (sustituirlo de vuelta, un caso límite)? | un resultado sin verificar es una opinión |
| ¿Los nombres dicen qué son? | `fuerza` se lee sola; `f`, `aux2` y `res` no |
| ¿La discusión física responde lo que se preguntó? | el código correcto con conclusión equivocada sigue estando mal |

## Cómo se escribe un comentario

Un buen comentario de revisión es **concreto**: señala la línea, dice qué pasa,
por qué importa y propone algo. Se refiere al código, no a la persona.

| En lugar de | Escribe |
|---|---|
| "Esto está mal." | "La celda 7 usa `solucion`, pero se define hasta la celda 9: con *Restart & Run All* falla. ¿La subes?" |
| "No se entiende." | "No sé qué guarda `aux2`. ¿Le pondrías un nombre como `energia_cinetica`?" |
| "¿Por qué hiciste esto?" | "¿Por qué `expand` antes de `solve`? Yo esperaba `factor`. Si hay una razón, un comentario en la celda ayudaría." |

Cuando el arreglo es una línea, GitHub permite proponerla tal cual con un
bloque **`suggestion`**. Quien recibe el comentario la acepta con un clic
(*Commit suggestion*) y se hace el commit por sí solo:

````markdown
`f` no dice qué es; con el nombre que pide el enunciado se lee sola:

```suggestion
fuerza = -sp.diff(energia_potencial, x)
```
````

Si es un detalle menor que no bloquea nada, dilo: "*Detalle:* ..." Así quien
recibe la revisión sabe qué es urgente y qué no.
