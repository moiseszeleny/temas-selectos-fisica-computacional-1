## Cómputo simbólico: SymPy vs. otros CAS

Mathematica, Maple o el *Symbolic Math Toolbox* de MATLAB son sistemas de
álgebra computacional muy usados en física, pero son software privativo.
SymPy ofrece lo mismo con un modelo distinto:

| Característica | SymPy | Mathematica / Maple / MATLAB Symbolic |
|---|---|---|
| Licencia | Libre (BSD), gratuita | Privativa, de pago (licencia institucional o personal) |
| Código fuente | Abierto — se puede leer y modificar | Cerrado |
| Ecosistema | Integrado con NumPy, SciPy, Matplotlib: todo Python | Ecosistema propio, interoperabilidad limitada con Python |
| Extensibilidad | Se subclasifica directamente en Python (`Function`, `Symbol`, `Expr`) | Extensible en su propio lenguaje, no vía herencia de clases Python |
| Portabilidad | Corre donde corra Python | Requiere instalación y licencia de la plataforma específica |

Que sea código abierto no es solo una cuestión de costo: en el Módulo 4 del
curso vamos a **leer y extender la arquitectura interna de SymPy** para crear
nuestros propios objetos matemáticos — algo que solo es posible porque el
código está ahí para inspeccionarlo.


## Demo motivacional

Esto es solo una demostración — **no se espera que ustedes escriban código
todavía**. La primera vez que abramos SymPy formalmente será en la semana 4.

Vamos a resolver, de forma simbólica, la ecuación de movimiento de un
oscilador armónico simple:

$$\ddot{x}(t) + \omega^2 x(t) = 0$$


```python
import sympy as sp

sp.init_printing()

t = sp.symbols('t', real=True)
omega = sp.symbols('omega', positive=True)
x = sp.Function('x')

ecuacion = sp.Eq(x(t).diff(t, 2) + omega**2 * x(t), 0)
solucion = sp.dsolve(ecuacion, x(t))
ecuacion, solucion
```

SymPy no solo resuelve la ecuación diferencial: también podemos pedirle que
verifique, de forma simbólica, que la solución es correcta.

```python
# Sustituimos la solución en la ecuación original y simplificamos
sp.simplify(ecuacion.lhs.subs(x(t), solucion.rhs).doit())
```
