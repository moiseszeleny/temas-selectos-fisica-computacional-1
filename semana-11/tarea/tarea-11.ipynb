## El código que vas a probar

Una partícula de masa $m$ en una caja de ancho $L$ (un pozo infinito entre
$x = 0$ y $x = L$) tiene niveles $n = 1, 2, 3, \dots$, con

$$\psi_n(x) = \sqrt{\frac{2}{L}}\,\sin\frac{n\pi x}{L},
\qquad E_n = \frac{n^2\pi^2\hbar^2}{2mL^2}.$$

La celda de abajo trae las dos funciones ya escritas, y el nivel $n$ se valida:
si no es un entero positivo, lanzan `ValueError`. **No la cambies**: tu trabajo
es escribir las pruebas que demuestren que están bien.

```python
import pytest
import sympy as sp

sp.init_printing()

# Ya declarados: no los cambies, los tests los usan tal cual.
x = sp.Symbol("x", real=True)                           # posición dentro de la caja
L, m, hbar = sp.symbols("L m hbar", positive=True)      # ancho, masa y hbar: positivos


def validar_nivel(n):
    """Lanza ValueError si n no es un entero positivo: n = 1, 2, 3, ..."""
    n = sp.sympify(n)
    if not (n.is_integer and n.is_positive):
        raise ValueError(f"El nivel debe ser un entero positivo, no {n}")


def funcion_de_onda(n, L, x):
    """Función de onda normalizada del nivel n de una partícula en una caja [0, L]."""
    validar_nivel(n)
    return sp.sqrt(2 / L) * sp.sin(n * sp.pi * x / L)


def energia_caja(n, m, L):
    """Energía del nivel n de una partícula de masa m en una caja de ancho L."""
    validar_nivel(n)
    return n**2 * sp.pi**2 * hbar**2 / (2 * m * L**2)
```

```python
display(sp.Eq(sp.Symbol("psi_2"), funcion_de_onda(2, L, x)))
display(sp.Eq(sp.Symbol("E_2"), energia_caja(2, m, L)))
```
